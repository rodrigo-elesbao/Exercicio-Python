```python
# Criei uma função chamada média de notas para aluno com 02 provas(nota 1 e nota2).

def media_notas(nota1,nota2):

# Função que vai somar as duas notas e calcular a média
    return (nota1+nota2)/2

# Vai abrir uma campo para digitar as notas, esse campos grava o valor das respectiva notas.
nota1 = float(input("Digite a nota1 :"))
nota2 = float(input("Digite a nota 2:"))

# Funcão que vai resolver a equação - soma das 2 notas e realizar a divisão, o resultado fica armazendo no "resultado final"
resultado_final = media_notas(nota1,nota2)

# Mostrar o resultado final da média do aluno
print(f"A média do aluno é:{resultado_final}")

```
